```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar el dataset
df = pd.read_csv("/Users/carlis/mi_proyecto/Visualizacion avanzada/World-happiness-report-2024.csv")

print("Primeras filas del DataFrame:")
print(df.head(10))

print("Ultimas filas del DataFrame:")
print(df.tail(5))

print("Informacion del DataFrame:")
print(df.info())

print("Estadisticas descriptivas del DataFrame:")
print(df.describe().round(2))
```

```python
df_numeric = df.select_dtypes(include='number')

# Calcular las correlaciones
correlation_matrix = df_numeric.corr(method='pearson')
spearman_corr = df_numeric.corr(method='spearman')
kendall_corr = df_numeric.corr(method='kendall')

# Crear mapas de calor
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title("Mapa de Calor de la Correlación de Pearson")
plt.show()

plt.figure(figsize=(10, 8))
sns.heatmap(spearman_corr, annot=True, cmap='viridis', center=0)
plt.title("Mapa de Calor de la Correlación de Spearman")
plt.show()

plt.figure(figsize=(10, 8))
sns.heatmap(kendall_corr, annot=True, cmap='plasma', center=0)
plt.title("Mapa de Calor de la Correlación de Kendall")
plt.show()
```

El analisis de correlacion de Pearson muestra relaciones positivas entre la mayoria de las variables asociadas al bienestar y felicidad de los paises. Las correlaciones mas fuertes se observan entre el Ladder Score y variables como social support, log GDP per capita y Healthy life expectancy, lo que sugiere que mejores condicion economicas, sociales y de salud se relacionan directamente con mayores niveles de felicidad.

En la correlacion de Spearman se confirma que variables como PIB per capita, apoyo social y esperanza de vida saludable presentan las relaciones mas fuertes con los niveles de felicidad, mientras que variables como generosidad muestran una influencia considerablemente menor dentro del dataset analizado.

El analisis de Kendall confirma que las variables economicas, sociales y de salud presentan asociaciones poistivas con los niveles de felicidad. Aunque las correlaciones son mas moderadas en comparacion con Pearson y Spearman, los resultados mantienen las mismas tendencias generales observadas en los analisis anteriores.

Estas correlaciones son relevantes para el analisis de datos y la toma de decisiones, ya que permiten identificar cuales factores tienen mayor relacion con el bienestar de los paises. Esta informacion puede ser util para orientar politicas publicas enfocadas en mejorar las condiciones economicas, sociales y de salud de la poblacion, contribuyendo asi a incrementar los niveles de felicidad y calidad de vida.
